```python
from google import genai
from google.genai import types
import os
from dotenv import load_dotenv
```

```python
load_dotenv(override=True)
```

```python
# API anahtarınızı buraya girin veya ortam değişkeninden alın
client = genai.Client()

# Modeli "google_search_retrieval" aracıyla yapılandırıyoruz
# Modeli Google Arama özelliğiyle çalıştırın
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Bugün iran savasi ile ilgili onemli haberler var mi? Varsa o haberleri ozetle.",
    config=types.GenerateContentConfig(
        tools=[types.Tool(google_search=types.GoogleSearchRetrieval())]
    )
)

# Yanıtı yazdıralım
print("-" * 30)
print("YANIT:")
print(response.text)

# Kaynakları (Grounding) görmek isterseniz:
if response.candidates[0].grounding_metadata:
    print("\nKAYNAKLAR:")
    for chunk in response.candidates[0].grounding_metadata.grounding_chunks:
        if chunk.web:
            print(f"- {chunk.web.title}: {chunk.web.uri}")
```
